# 2. Conditional Statements

Conditional statements are a fundamental concept in programming that allow a program to **make decisions** and execute different blocks of code based on certain conditions.

---

### **1. What are Conditional Statements?**

A conditional statement checks whether a condition is **True or False**.

* If the condition is **True**, a specific block of code is executed.
* If the condition is **False**, another block (or no block) is executed.

They help control the **flow of a program**.

---

### **2. Types of Conditional Statements in Python**

#### **a) `if` Statement**

* Executes code only if the condition is true.

**Syntax:**

```python
if condition:
    # code to execute
```

---

#### **b) `if-else` Statement**

* Provides an alternative block if the condition is false.

**Syntax:**

```python
if condition:
    # code if true
else:
    # code if false
```

---

#### **c) `if-elif-else` Statement**

* Used when there are multiple conditions to check.

**Syntax:**

```python
if condition1:
    # code
elif condition2:
    # code
else:
    # code
```

---

### **3. Logical Operators in Conditions**

Conditional statements often use logical operators:

* **`and`** → True only if both conditions are true
* **`or`** → True if at least one condition is true
* **`not`** → Reverses the condition

---

### **Conclusion**

Conditional statements are essential for controlling program execution. They enable a program to **respond differently to different inputs**, making it more flexible and intelligent.


````python
print("Enter the amount of your product")
amount = int(input())

if amount > 400:
  amount = (amount-(amount*10/100))
  print("You have to pay", amount)

elif amount > 200 or amount < 400:
  amount = (amount-(amount*5/100))
  print("You have to pay", amount)
  
else:
  print("Pay amount", amount)
````

### **Loops / Iteration**

Loops (or iteration) are used in programming to **repeat a block of code multiple times** without writing it again and again. They help make programs shorter, more efficient, and easier to understand.

---

### **1. What is Iteration?**

Iteration means **executing the same set of instructions repeatedly** until a condition is met or all items in a collection are processed.

---

### **2. Types of Loops in Python**

#### **a) `for` Loop**

* Used to iterate over a sequence such as a list, tuple, or string.
* It runs once for each item in the sequence.

**Syntax:**

```python
for variable in sequence:
    # code block
```

---

#### **b) `while` Loop**

* Repeats as long as a condition is **True**.

**Syntax:**

```python
while condition:
    # code block
```

---


---

#### **c) Final Output**

```python
print(total)
```

* Prints the final sum after the loop finishes.

---

### **3. Importance of Loops**

* Avoid repetition of code
* Process large amounts of data easily
* Perform calculations like totals, averages, etc.

---

### **Conclusion**

Loops are essential for performing repetitive tasks. In your program, the `for` loop is used to **iterate through a list and calculate the total sum**, demonstrating how iteration simplifies data processing.


````python
transaction = [100, 200, 300, 400, 500]

total = 0

for t in transaction:
    total = total+t
    print("Total is", total)


print(total)
````

### **Functions in Python**

Functions are an important concept in programming that allow you to **organize code into reusable blocks**. Instead of writing the same code multiple times, you can define a function once and use it whenever needed.

---

### **1. What is a Function?**

A function is a **named block of code** that performs a specific task.
It runs only when it is **called (invoked)**.

---

### **2. Why Use Functions?**

* Reduce repetition (**code reusability**)
* Make programs more **organized and readable**
* Break complex problems into **smaller parts**
* Easier to test and debug

---

### **3. Defining a Function**

In Python, functions are defined using the `def` keyword.

**Syntax:**

```python
def function_name(parameters):
    # code block
    return value
```

---

### **4. Calling a Function**

To use a function, you must call it by its name:

```python
function_name(arguments)
```

---

### **5. Types of Functions**

#### **a) Function without Parameters**

```python
def greet():
    print("Hello")

greet()
```

---

#### **b) Function with Parameters**

```python
def greet(name):
    print("Hello", name)

greet("John")
```

---

#### **c) Function with Return Value**

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

* `return` sends the result back to where the function was called.

---

### **6. Key Concepts**

#### **a) Parameters vs Arguments**

* **Parameters** → variables in the function definition
* **Arguments** → values passed when calling the function

---

#### **b) Local Scope**

* Variables inside a function exist only within that function.

---

#### **c) Reusability**

* A function can be used multiple times with different inputs.

---

### **7. Importance of Functions**

* Help in **modular programming**
* Improve **code clarity**
* Save time and effort
* Make large programs easier to manage

---

### **Conclusion**

Functions are a powerful tool in Python that allow you to **structure your code efficiently**. They make programs more reusable, readable, and easier to maintain by breaking tasks into smaller, manageable pieces.
